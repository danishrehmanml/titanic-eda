#import library and upload data set

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)
````

## info function to check columns detail datatypes and null values and shape attribute to check diemension

````python
df.info()   # info about columns
print("shape")
df.shape  # telling about rows and column
````

````python
df.describe()   #telling us about each column mean min max standard devation
````

````python
df.head() # by default give 1st 5 lines
````

````python
df.isna().sum()  # column wise null values
````

````python
df.isna().sum().sum()  # sum all null values in dataset
````

````python
df=df.drop(["Cabin"],axis=1)   #droing cabin column due to a more than 50 percen null value
````

````python
df
````

````python
df.isna().sum().sum()  # sum all null values in dataset
````

````python
df["Age"]=df["Age"].fillna(df["Age"].mean())  #filin age column because non null value is more than 50% so through mean null value is being filled
````

````python
df.isna().sum().sum()  # sum all null values in dataset
````

````python
df["Embarked"]=df["Embarked"].fillna(df["Embarked"].mode()[0])  #Embarked is being filled because only 2 null value is present filied by mode because Embarked filed by most frequent value
````

````python
df.isna().sum().sum()  # sum all null values in dataset
````

````python
df=df.drop(["Name","Ticket","PassengerId"],axis=1)
````

````python
df.shape
````

````python
df.isnull().sum()
````

````python
print(df[df["Fare"]==0])
````

````python
df=df[df["Fare"]!=0]
df
````

````python
print(df.groupby("Embarked")["Survived"].mean())
````

````python
df["Agegroup"]=pd.cut(df["Age"],
                      bins=[0,12,24,48,60,100],
                      labels=["child","teen","young","mature","old"])
print(df.groupby("Agegroup")["Survived"].mean())
````

## Age Group Analysis
- Children had highest survival (57%)
- Seniors had lowest survival (22%)
- Middle aged (45-60) survived more than young adults
